## **Exercício 20 - Matriz Transposta**
Dada uma matriz (lista de listas), crie sua matriz transposta.

```python

# Cria uma matriz 3x3.
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Cria uma lista para armazenar a matriz transposta.
transposta = []

# Percorre as colunas da matriz.
for coluna in range(len(matriz[0])):

    # Cria uma nova linha para a matriz transposta.
    nova_linha = []

    # Percorre as linhas da matriz original.
    for linha in range(len(matriz)):

        # Adiciona o elemento correspondente à nova linha.
        nova_linha.append(matriz[linha][coluna])

    # Adiciona a nova linha à matriz transposta.
    transposta.append(nova_linha)

# Exibe a matriz original.
print("Matriz original:")
for linha in matriz:
    print(linha)

# Exibe a matriz transposta.
print("\nMatriz transposta:")
for linha in transposta:
    print(linha)
```

## **Exercício 21 = Busca Binária**
Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

```python

# Cria uma lista de números em ordem crescente.
numeros = [2, 5, 8, 12, 16, 21, 25, 30, 35]

# Solicita o número que será procurado.
numero = int(input("Digite o número que deseja procurar: "))

# Define o início e o fim da área de busca.
inicio = 0
fim = len(numeros) - 1

# Indica se o número foi encontrado.
encontrado = False

# Continua a busca enquanto houver elementos para verificar.
while inicio <= fim:

    # Calcula a posição do elemento central.
    meio = (inicio + fim) // 2

    # Verifica se o elemento central é o número procurado.
    if numeros[meio] == numero:
        encontrado = True
        break

    # Se o número procurado for maior, procura na metade direita.
    elif numero > numeros[meio]:
        inicio = meio + 1

    # Se for menor, procura na metade esquerda.
    else:
        fim = meio - 1

# Exibe o resultado da busca.
if encontrado:
    print(f"{numero} foi encontrado na lista.")
else:
    print(f"{numero} não foi encontrado na lista.")
```

## **Exercício 22 - Criptografia de Cifra de César**

Crie uma função para criptografar uma mensagem cifrada com a
cifra de César.



```python


mensagem = "Vai, Corinthians!!"
chave = 3

# Cria uma função para criptografar a mensagem utilizando a Cifra de César.
def criptografiaCesar(texto, chave):
    texto_cifrado = ""

    # Percorre cada caractere da mensagem.
    for caractere in texto:

        # Verifica se o caractere é uma letra.
        if caractere.isalpha():

            # Verifica se a letra está em maiúscula.
            if caractere.isupper():
                texto_cifrado += chr((ord(caractere) - 65 + chave) % 26 + 65)

            # Caso contrário, mantém a letra em minúscula.
            else:
                texto_cifrado += chr((ord(caractere) - 97 + chave) % 26 + 97)

        # Mantém espaços e caracteres especiais sem alteração.
        else:
            texto_cifrado += caractere

    return texto_cifrado


# Utiliza a função para criptografar a mensagem.
mensagem_cifrada = criptografiaCesar(mensagem, chave)

# Exibe a mensagem criptografada.
print(f"Mensagem cifrada: {mensagem_cifrada}")
```

## **Exercício 24 - Validação de CPF**

Implemente uma função para validar um CPF brasileiro.





```python
import re

# Cria uma função que valida CPF.
def validaCpf(cpf):

    # Verifica se o CPF está em um dos formatos aceitos.
    padrao = re.compile(r'^\d{3}\.\d{3}\.\d{3}-\d{2}$|^\d{11}$')

    # Verifica se o CPF segue o formato esperado.
    if not re.match(padrao, cpf):
        return False

    # Remove pontos e hífen para realizar os cálculos.
    cpf = cpf.replace(".", "").replace("-", "")

    # Calcula o primeiro dígito verificador.
    soma = 0

    for i in range(9):
        soma += int(cpf[i]) * (10 - i)

    resto = soma % 11

    if resto < 2:
        primeiro_digito = 0
    else:
        primeiro_digito = 11 - resto

    # Verifica se o primeiro dígito está correto.
    if primeiro_digito != int(cpf[9]):
        return False

    # Calcula o segundo dígito verificador.
    soma = 0

    for i in range(10):
        soma += int(cpf[i]) * (11 - i)

    resto = soma % 11

    if resto < 2:
        segundo_digito = 0
    else:
        segundo_digito = 11 - resto

    # Verifica se o segundo dígito está correto.
    if segundo_digito != int(cpf[10]):
        return False

    return True


# Solicita o CPF ao usuário.
cpf = input("Digite um CPF: ")

# Exibe o resultado da validação.
if validaCpf(cpf):
    print("CPF válido.")
else:
    print("CPF inválido.")
```

## **Exercício 30 - Jogo da Forca**
Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.



```python
import random

# Lista de palavras que podem ser escolhidas pelo computador.
palavras = ["python", "java", "programacao", "algoritmo"]

# Escolhe uma palavra aleatoriamente.
palavra = random.choice(palavras)

# Cria uma lista com "_" para esconder as letras da palavra.
letras_descobertas = ["_"] * len(palavra)

# Define a quantidade máxima de erros.
erros = 0
limite_erros = 6

# Continua o jogo enquanto o jogador não acertar a palavra
# e ainda tiver tentativas disponíveis.
while "_" in letras_descobertas and erros < limite_erros:

    # Exibe o estado atual da palavra.
    print("Palavra:", " ".join(letras_descobertas))

    # Solicita uma letra ao jogador.
    letra = input("Digite uma letra: ").lower()

    # Verifica se a letra está presente na palavra.
    if letra in palavra:

        # Percorre cada posição da palavra.
        for i in range(len(palavra)):

            # Verifica se a letra digitada está naquela posição.
            if palavra[i] == letra:

                # Revela a letra acertada na posição correspondente.
                letras_descobertas[i] = letra

        print("Você acertou a letra!")

    else:
        # Aumenta a quantidade de erros.
        erros += 1
        print(f"Você errou! Erros: {erros}/{limite_erros}")

# Verifica se o jogador conseguiu descobrir todas as letras.
if "_" not in letras_descobertas:
    print(f"Parabéns! Você acertou a palavra: {palavra}")
else:
    print(f"Você perdeu! A palavra era: {palavra}")
```

## **Exercício 31 - Ordenação por Bolha**

Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
# Cria uma lista de números desordenados.
numeros = [64, 34, 25, 12, 22, 11, 90]

# Percorre a lista várias vezes para realizar as comparações.
for i in range(len(numeros)):

    # Compara os elementos vizinhos da lista.
    for j in range(len(numeros) - 1):

        # Verifica se o elemento atual é maior que o próximo.
        if numeros[j] > numeros[j + 1]:

            # Troca os elementos de posição para ordená-los.
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

# Exibe a lista ordenada.
print(f"Lista ordenada: {numeros}")
```

```python
# Exercício - Jogo da Velha

# Cria o tabuleiro com 9 posições vazias.
tabuleiro = [" ", " ", " ",
             " ", " ", " ",
             " ", " ", " "]

# Define o jogador que começa a partida.
jogador = "X"

# Controla a quantidade de jogadas realizadas.
jogadas = 0

# Continua o jogo enquanto houver posições disponíveis.
while jogadas < 9:

    # Exibe o tabuleiro atual.
    print(f"""
     {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]}
    ---+---+---
     {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]}
    ---+---+---
     {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]}
    """)

    # Solicita ao jogador a posição onde deseja jogar.
    posicao = int(input(f"Jogador {jogador}, escolha uma posição de 1 a 9: ")) - 1

    # Verifica se a posição escolhida está disponível.
    if tabuleiro[posicao] == " ":

        # Coloca o símbolo do jogador na posição escolhida.
        tabuleiro[posicao] = jogador
        jogadas += 1

        # Verifica todas as possibilidades de vitória.
        if (tabuleiro[0] == jogador and tabuleiro[1] == jogador and tabuleiro[2] == jogador) or \
           (tabuleiro[3] == jogador and tabuleiro[4] == jogador and tabuleiro[5] == jogador) or \
           (tabuleiro[6] == jogador and tabuleiro[7] == jogador and tabuleiro[8] == jogador) or \
           (tabuleiro[0] == jogador and tabuleiro[3] == jogador and tabuleiro[6] == jogador) or \
           (tabuleiro[1] == jogador and tabuleiro[4] == jogador and tabuleiro[7] == jogador) or \
           (tabuleiro[2] == jogador and tabuleiro[5] == jogador and tabuleiro[8] == jogador) or \
           (tabuleiro[0] == jogador and tabuleiro[4] == jogador and tabuleiro[8] == jogador) or \
           (tabuleiro[2] == jogador and tabuleiro[4] == jogador and tabuleiro[6] == jogador):

            print(f"Jogador {jogador} venceu!")
            break

        # Alterna entre os jogadores X e O.
        if jogador == "X":
            jogador = "O"
        else:
            jogador = "X"

    else:
        # Informa que a posição escolhida já está ocupada.
        print("Essa posição já está ocupada.")

# Verifica se todas as posições foram preenchidas sem um vencedor.
if jogadas == 9:
    print("O jogo terminou em empate.")
```

## **Exercício 37 - Funções Recursivas**
Implemente uma função recursiva para calcular o fatorial de um
número.

```python
# Solicita um número ao usuário.
numero = int(input("Digite um número: "))


# Cria uma função recursiva para calcular o fatorial.
def fatorial(numero):

    # Define o caso base da recursão.
    if numero == 0:
        return 1

    # Chama a própria função para calcular o próximo valor.
    else:
        return numero * fatorial(numero - 1)


# Utiliza a função para calcular o fatorial do número informado.
resultado = fatorial(numero)

# Exibe o resultado do cálculo.
print(f"O fatorial de {numero} é {resultado}.")
```

## **Exercício 40 - Lista de Compras**

Crie um programa para gerenciar uma lista de compras, permitindo
adicionar, remover e listar itens.

```python

# Cria uma lista vazia para armazenar os itens.
lista_compras = []

# Mantém o programa funcionando até o usuário escolher sair.
while True:

    # Exibe o menu de opções.
    print("\n1 - Adicionar item")
    print("2 - Remover item")
    print("3 - Listar itens")
    print("4 - Sair")

    # Solicita ao usuário uma opção.
    opcao = input("Escolha uma opção: ")

    # Verifica a opção escolhida pelo usuário.
    match opcao:

        # Adiciona um item à lista de compras.
        case "1":
            item = input("Digite o item que deseja adicionar: ")
            lista_compras.append(item)
            print("Item adicionado com sucesso.")

        # Remove um item da lista de compras.
        case "2":
            item = input("Digite o item que deseja remover: ")

            if item in lista_compras:
                lista_compras.remove(item)
                print("Item removido com sucesso.")
            else:
                print("Esse item não está na lista.")

        # Exibe todos os itens da lista.
        case "3":
            if lista_compras:
                print("Lista de compras:")

                for item in lista_compras:
                    print(f"- {item}")
            else:
                print("A lista de compras está vazia.")

        # Encerra o programa.
        case "4":
            print("Programa encerrado.")
            break

        # Trata opções que não existem no menu.
        case _:
            print("Opção inválida.")
```

## **Exercício 42 -  Números Primos até N**
Gere todos os números primos até um número n fornecido.

```python

# Solicita um número ao usuário.
n = int(input("Digite um número: "))

# Percorre todos os números de 2 até o número informado.
for numero in range(2, n + 1):

    # Considera o número como primo inicialmente.
    numero_primo = True

    # Verifica se o número possui algum divisor além de 1 e ele mesmo.
    for divisor in range(2, numero):

        # Se houver divisão exata, o número não é primo.
        if numero % divisor == 0:
            numero_primo = False
            break

    # Exibe o número caso ele seja primo.
    if numero_primo:
        print(numero)
```

## **Exercício 47 - Contar Números em Lista**
Conte quantas vezes cada número aparece em uma lista.

```python
# Cria uma lista com números repetidos.
numeros = [1, 2, 2, 3, 3, 3, 4, 5, 5]

# Cria um dicionário vazio para armazenar a quantidade de vezes
# que cada número aparece.
contagem = {}

# Percorre cada número da lista.
for numero in numeros:

    # Verifica se o número já existe no dicionário.
    if numero in contagem:

        # Aumenta a quantidade do número em 1.
        contagem[numero] += 1

    # Caso o número ainda não exista no dicionário.
    else:

        # Adiciona o número ao dicionário começando sua contagem em 1.
        contagem[numero] = 1

# Exibe a quantidade de vezes que cada número aparece.
print(f"Contagem dos números: {contagem}")
```

## **Exercício 48 - Ordenação por Inserção**

Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
# Cria uma lista de números desordenados.
numeros = [5, 2, 4, 1, 3]

# Percorre a lista a partir do segundo elemento.
for i in range(1, len(numeros)):

    # Armazena o número atual.
    numero_atual = numeros[i]

    # Define a posição anterior ao número atual.
    posicao = i - 1

    # Percorre os elementos anteriores enquanto forem maiores
    # que o número atual.
    while posicao >= 0 and numeros[posicao] > numero_atual:

        # Move o número maior uma posição para a direita.
        numeros[posicao + 1] = numeros[posicao]

        # Volta uma posição na lista.
        posicao -= 1

    # Coloca o número atual na posição correta.
    numeros[posicao + 1] = numero_atual

# Exibe a lista ordenada.
print(f"Lista ordenada: {numeros}")
```

## **Exercício 50 - Cálculo de Frequência de Letras**

Calcule a frequência de cada letra em uma string fornecida pelo
usuário.

```python
# Solicita uma frase ao usuário.
frase = input("Digite uma frase: ")

# Cria um dicionário vazio para armazenar a frequência das letras.
frequencia = {}

# Percorre cada caractere da frase.
for letra in frase:

    # Verifica se o caractere é uma letra.
    if letra.isalpha():

        # Converte a letra para minúscula.
        letra = letra.lower()

        # Verifica se a letra já existe no dicionário.
        if letra in frequencia:

            # Aumenta a quantidade da letra em 1.
            frequencia[letra] += 1

        # Caso a letra ainda não exista no dicionário.
        else:

            # Adiciona a letra ao dicionário começando sua contagem em 1.
            frequencia[letra] = 1

# Exibe a frequência de cada letra.
print(f"Frequência das letras: {frequencia}")
```
