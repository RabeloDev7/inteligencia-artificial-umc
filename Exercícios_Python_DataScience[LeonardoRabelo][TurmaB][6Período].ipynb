19 - Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

4 - Use a função de otimização de SciPy para maximizar f(x) = −x^2+4x+1

```python
from scipy.optimize import minimize_scalar


def funcao(x):
  return -x**2 +4*x +1

# scipy só minimiza, então maximizar f(x) equivale a minimizar -f(x).
def funcao_negativa(x):
  return -funcao(x)

# Encontra o x que minimiza funcao_negativa (ou seja, que maximiza f).
resultado = minimize_scalar(funcao_negativa)

# x onde ocorre o máximo.
x_max = resultado.x

# Valor de f da função original.
valor_max = funcao(x_max)

# Exibe os resultados de x e valor máximo.
print(f"X que maximiza: {x_max:.2f}")
print(f"Valor máximo: {valor_max:.2f}")

```

5 - Calcule a transformada de Fourier de um sinal composto de sin(2πt) e
cos(4πt).

```python
from scipy.fft import fft, fftfreq
import numpy as np

# Cria o eixo do tempo (500 pontos entre 0 e 1 segundo)
t = np.linspace(0, 1, 500)

# Sinal composto: soma de duas ondas (1 Hz e 2 Hz)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

# Calcula a transformada de Fourier do sinal
frequencias = fft(sinal)

# Gera o eixo de frequências correspondente a cada valor da transformada
eixo_freq = fftfreq(len(t), t[1] - t[0])

# Mostra só as frequências com amplitude relevante
for f, amp in zip(eixo_freq, np.abs(frequencias)):
    if f >= 0 and amp > 10:
        print(f"Frequência: {f:.1f} Hz, Amplitude: {amp:.2f}")
```

para x e [0,1] para y.

8 - Plote um gráfico de linha com intervalos de confiança (sombramento) para
a função y = sin(x).

```python
import matplotlib.pyplot as plt
import numpy as np

# Cria os valores de x.
x = np.linspace(0, 10, 100 )

# Cria os valores de y.
y = np.sin(x)

# Define uma margem de erro de 0.2.
erro = 0.2

# Define o limite inferior.
limite_inferior = y - erro

# Define o limite superior.
limite_superior = y + erro

# Cria a linha principal.
plt.plot(x, y, label="sin(x)")

# Cria o intervalor sombreado.
plt.fill_between(
    x,
    limite_inferior,
    limite_superior,
    alpha=0.3,
    label="Intervalo"
)

# Adiciona as informações.
plt.title("Função Seno com Intervalo Sombreado")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()

# Exibe o gráfico sombreado.
plt.show()
```
